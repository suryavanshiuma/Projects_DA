#### Parametric Insurance Case Study

Solar Generation Risk Assessment and Insurance Pricing Using Historical Weather Data

#### Executive Summary

This case study develops a simple parametric insurance model for a residential rooftop solar system using historical weather data from 2005 to 2024. Solar energy generation was estimated from ERA5-Land solar radiation data for Ahmedabad (Pincode 380006). Historical generation was then back-tested against a predefined annual energy threshold to identify insurance payout years. Finally, the burn cost method was used to estimate the expected annual loss and recommend a commercial insurance premium.

#### Problem Statement

The objective of this case study is to develop a simple parametric insurance model for a residential rooftop solar PV system. Using historical weather data, the model estimates annual solar energy generation, identifies years with generation shortfalls, and calculates insurance payouts based on a predefined trigger value. The study also estimates an appropriate commercial premium using the burn cost pricing approach.

#### Methodology

The analysis was carried out in the following steps:

1. Converted ERA5-Land NetCDF weather data into CSV format.
2. Filtered data for the selected location (Ahmedabad - Pincode 380006).
3. Converted solar radiation (SSRD) into kWh/m²/day.
4. Estimated monthly and annual solar energy generation.
5. Applied a parametric insurance trigger of **4,600 kWh/year**.
6. Calculated annual payouts for years with generation shortfalls.
7. Estimated the expected annual loss (Burn Cost) and recommended a commercial insurance premium.

### Task 1 – Weather Index Selection

#### Selected Weather Variable

The selected weather variable for this study is **Surface Solar Radiation Downwards (SSRD)** obtained from the ERA5-Land dataset. SSRD directly measures the amount of solar energy reaching the Earth's surface, making it the most suitable variable for estimating electricity generation from rooftop solar photovoltaic (PV) systems.

#### Why SSRD Was Selected

SSRD was selected because electricity generation from solar PV systems primarily depends on the amount of solar radiation received by the panels. Higher solar radiation generally results in higher electricity generation, while lower radiation reduces energy production. Therefore, SSRD provides the most direct relationship with annual solar energy generation and minimizes basis risk when designing the parametric insurance product.

#### Why Other Weather Variables Were Not Selected

Other weather variables such as rainfall, temperature, and cloud cover were considered but were not selected as the primary index. These variables may indirectly influence solar power generation; however, they do not measure the amount of solar energy available to the PV panels. Since the objective is to estimate electricity generation accurately, SSRD provides the most direct and reliable weather index.

#### Assumptions

The analysis assumes a representative **3 kVA rooftop solar PV system** with a **Performance Ratio (PR) of 80%**. Historical weather conditions are assumed to be representative of future climate patterns, and non-weather factors such as equipment failure, shading, maintenance issues, and panel degradation are not considered in the insurance model.

### Task 2 – Parametric Insurance Product Design

#### Weather Index

The insurance product uses **annual solar energy generation**, estimated from **Surface Solar Radiation Downwards (SSRD)**, as the weather index. Annual generation is calculated from historical weather data and compared against a predefined trigger value to determine whether a payout is required.

#### Trigger and Payout Structure

The insurance trigger is set at **4,600 kWh/year (AEP50)**. If the estimated annual generation falls below this threshold, the policy pays compensation based on the energy shortfall multiplied by the electricity tariff (**₹5.75/kWh**).

**Payout Formula:**

**Payout = Max(0, Trigger − Annual Generation) × Tariff**

#### Maximum Payout

The maximum payout occurs when annual solar energy generation is significantly below the trigger level. In this model, the payout is proportional to the energy shortfall, ensuring that compensation reflects the estimated financial loss due to reduced electricity generation.

#### Assumptions and Limitations

The product assumes that weather is the primary driver of changes in solar energy generation. Factors such as equipment failure, poor maintenance, shading, inverter losses beyond the assumed performance ratio, and customer operational issues are not covered. As a result, the product may not compensate for generation losses caused by non-weather-related events.

#### Results and Analysis

Historical solar generation was estimated for the period **2005–2024** using ERA5-Land weather data. Annual energy generation was compared with the predefined insurance trigger of **4,600 kWh/year** to identify years with generation shortfalls.

The historical back-testing results showed:

- **Analysis Period:** 2005–2024 (20 Years)
- **Average Annual Generation:** 4,566.88 kWh
- **Highest Annual Generation:** 4,826.44 kWh (2009)
- **Lowest Annual Generation:** 3,333.87 kWh (2023)
- **Insurance Trigger Activated:** 8 out of 20 years

The analysis indicates that annual solar generation remained close to the trigger level in most years. However, lower solar radiation during certain years resulted in generation shortfalls, activating insurance payouts. The most significant shortfall occurred in **2023**, while **2009** recorded the highest annual generation with no payout required.

### Insurance Pricing

The insurance product was priced using the **Burn Cost Method**, which estimates the expected annual loss from historical payout data. A commercial premium was then calculated by adding administrative and risk loadings to the expected annual loss.

#### Pricing Methodology

**Estimated Annual Generation**

Annual Generation = Σ (Monthly Solar Generation)

**Insurance Payout**

Payout = Max (0, Trigger − Annual Generation) × Tariff

**Burn Cost**

Burn Cost = Average Historical Annual Payout

**Commercial Premium**

Commercial Premium = Burn Cost × (1 + Administrative Loading + Risk Loading)

#### Pricing Assumptions

- **Administrative Loading:** 20%
- **Risk Loading:** 15%

The calculated commercial premium provides an estimate of the annual premium required to cover expected claims, operational expenses, and insurer risk.

### Task 4 – Product Critique

Although the proposed parametric insurance product is simple, transparent, and easy to implement, it has certain limitations when applied to a large customer base. If this product is offered to 10,000 customers, the following situations may arise.

#### 1. Basis Risk

Customers may experience financial losses even when the insurance trigger is not activated. For example, losses caused by equipment failure, shading, or poor maintenance are not reflected in the weather index.

**Improvement:** Combine weather-based triggers with system performance data to reduce basis risk.

#### 2. Extreme Weather Events

Future weather conditions may differ from historical patterns due to climate change. As a result, historical data alone may not accurately represent future insurance risk, leading to higher-than-expected payouts.

**Improvement:** Regularly update the pricing model using recent weather data and periodically review the trigger level to reflect changing climate conditions.

#### 3. Single Trigger Limitation

Using a single annual generation trigger may not capture all variations in solar power generation across different locations or customer installations. Customers with similar weather conditions may still experience different levels of electricity generation.

**Improvement:** Develop location-specific products and consider additional weather variables or system characteristics to improve pricing accuracy and reduce basis risk.

### Conclusion

This case study demonstrates how historical weather data can be used to design a simple parametric insurance product for rooftop solar systems. By estimating annual solar generation, applying a predefined trigger, and back-testing historical performance, the analysis identified years with energy shortfalls and corresponding insurance payouts. The Burn Cost method provided a practical approach for estimating a commercial insurance premium based on historical risk.

### Business Recommendation

The proposed parametric insurance model offers a transparent and data-driven approach to protecting solar asset owners against weather-related generation losses. Future improvements could include using multiple weather variables, evaluating different trigger levels, and testing the model across additional locations to enhance pricing accuracy and product reliability.
